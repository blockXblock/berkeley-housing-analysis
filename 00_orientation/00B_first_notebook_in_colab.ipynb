## Step 1: Import Our Tools

First, we load the Python libraries we'll use:
- **pandas**: For working with data tables (like Excel in Python)
- **matplotlib**: For making charts

```python
# Import pandas - we call it 'pd' for short
# This is like opening Excel, but more powerful
import pandas as pd

# Import matplotlib for making charts
# We call it 'plt' for short
import matplotlib.pyplot as plt

# This line makes charts show up right in the notebook
%matplotlib inline

print("Tools loaded successfully!")
```

## Step 2: Create a Toy Dataset

Let's create a small dataset of imaginary housing projects. In real life, we'd load this from a file, but creating it ourselves helps us understand the structure.

```python
# Create a dictionary with our data
# Each key is a column name, each value is a list of data
data = {
    'address': [
        '123 Main Street',      # Project 1
        '456 Oak Avenue',       # Project 2
        '789 Cedar Lane',       # Project 3
        '101 Pine Road',        # Project 4
        '202 Elm Court'         # Project 5
    ],
    'units': [
        45,   # Project 1 has 45 apartments
        12,   # Project 2 has 12 units
        8,    # Project 3 has 8 units
        150,  # Project 4 is a big building!
        6     # Project 5 is small
    ],
    'status': [
        'Under Review',      # Waiting for city approval
        'Approved',          # Got approval, waiting for permit
        'Under Construction',# Being built right now
        'Under Review',      # Still being reviewed
        'Completed'          # Done! People live there now
    ]
}

# Convert the dictionary to a DataFrame (a data table)
df = pd.DataFrame(data)

# Show the data
print("Here's our housing projects data:")
print("="*50)
df
```

## Step 3: Explore the Data

Let's ask some questions about our data.

```python
# Question 1: How many projects do we have?
num_projects = len(df)
print(f"Number of projects: {num_projects}")

# Question 2: How many total units?
total_units = df['units'].sum()
print(f"Total housing units: {total_units}")

# Question 3: What's the average project size?
avg_units = df['units'].mean()
print(f"Average units per project: {avg_units:.1f}")

# Question 4: What's the biggest project?
max_units = df['units'].max()
print(f"Largest project: {max_units} units")
```

```python
# Let's count projects by status
# This is called "grouping" - we group projects by their status

status_counts = df.groupby('status').size()
print("Projects by status:")
print(status_counts)
```

```python
# More useful: how many UNITS are in each status?
# This tells us how much housing is at each stage

units_by_status = df.groupby('status')['units'].sum()
print("Housing units by status:")
print(units_by_status)
```

## Step 4: Make a Bar Chart

Now let's visualize the data. A bar chart is perfect for comparing categories.

```python
# Create a figure (the canvas) and axes (where we draw)
fig, ax = plt.subplots(figsize=(10, 6))

# Create a bar chart of units by status
# We use different colors for each bar
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
units_by_status.plot(kind='bar', ax=ax, color=colors)

# Add labels and title
ax.set_xlabel('Status', fontsize=12)
ax.set_ylabel('Number of Housing Units', fontsize=12)
ax.set_title('Housing Units by Project Status', fontsize=14, fontweight='bold')

# Rotate x-axis labels so they're easier to read
plt.xticks(rotation=45, ha='right')

# Add value labels on top of each bar
for i, v in enumerate(units_by_status):
    ax.text(i, v + 2, str(v), ha='center', fontweight='bold')

# Adjust layout so nothing gets cut off
plt.tight_layout()

# Show the chart!
plt.show()

print("\nWhat does this chart tell us?")
print("- Most units (195) are still 'Under Review' - waiting for approval")
print("- Only 6 units have been completed")
print("- This is a 'toy' dataset - real Berkeley has thousands of units!")
```
