This notebook imports and cleans Berkeley housing permit data.

```python
# Import libraries
import pandas as pd
import numpy as np
import os

print('✅ Libraries imported')
```

## 1. Load Permit Data

Load raw permit data from Berkeley Planning Department.

```python
# Load permit data
# NOTE: Replace with your actual permit data file
permit_file = "/Users/johngage/berkeley-data/housing_projects_final_complete.csv"
#'/Users/johngage/berkeley-data/permits.csv'

if os.path.exists(permit_file):
    df_permits = pd.read_csv(permit_file)
    print(f'Loaded {len(df_permits):,} permits')
    print(f'Columns: {df_permits.columns.tolist()}')
else:
    print('❌ Permit file not found')
    print('Expected location:', permit_file)
```

```python
print(f'permits: '{len(df_permits):m}')
```

## 2. Filter for Housing Projects

Keep only building permits and zoning permits (exclude occupancy permits).

```python
# Filter permit types
# Adjust column names based on your data

housing_types = ['Building Permit', 'Zoning Permit', 'Occupancy Permit']

# Example filter - adjust to your column names:
df_housing = df_permits[df_permits['permits'].isin(housing_types)]

print(f'Housing projects: {len(df_housing):,}')
```

## 3. Calculate Net New Units

Calculate net new housing units (new units - demolished units).

```python
# Calculate net units
# Adjust based on your column names

df_housing['net_units'] = (
    df_housing['new_units'].fillna(0) - 
    df_housing['demolished_units'].fillna(0)
)

# Filter for positive net units
df_housing = df_housing[df_housing['net_units'] > 0]

print(f'Projects with net new units: {len(df_housing):,}')
print(f'Total net new units: {df_housing["net_units"].sum():,.0f}')
```

## 4. Clean Address Data

Standardize address format.

```python
# Clean and standardize addresses
df_housing['address_clean'] = (
    df_housing['address']
    .str.upper()
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
)

print('Sample addresses:')
print(df_housing['address_clean'].head(10).tolist())
```

## 5. Save Clean Data

Save cleaned housing projects.

```python
# Save clean data
output_file = '/Users/johngage/berkeley-data/housing_projects_clean.csv'
df_housing.to_csv(output_file, index=False)

print(f'✅ Saved {len(df_housing):,} projects to:')
print(f'   {output_file}')
```
