## What's in these files

The **geometry** file contains 180 housing projects, each rendered as a 3D extruded building footprint at its actual Berkeley location, with an address · units · stage label floating at roof height. The **tour** files (Over-200, Dormitory, Adeline & Shattuck, Elmwood) contain camera flythroughs that visit specific subsets of those projects.

**Color coding by stage:**
- Purple: UC Berkeley student housing
- Orange: Approved (entitled) but not yet permitted
- Yellow: At least one building permit issued, construction not yet started
- Blue: Under construction
- Green: Completed (Certificate of Occupancy issued)

**Building heights:** Reflect proposed or built height in stories. A 26-story tower extrudes to roughly 85 meters; a 4-story building to about 13 meters.

**Click any building** in Google Earth to see its details: address, unit count, developer, architect, key permit dates, and current status. Pause a tour during playback to read the labels on buildings as the camera passes.

**Hand-traced footprints:** Most buildings use approximate footprints derived from parcel boundaries. About a dozen of the most consequential projects (Modera Acheson Commons, 2400 Bowditch, 1950 Oxford, 2200 Bancroft) have hand-traced precise footprints based on architectural plans and aerial imagery.

## Data sources and accuracy

**Primary sources:**
- Berkeley Accela permit portal (live, updated as permits change)
- Planning Commission and Zoning Adjustments Board staff reports
- City Council agendas and resolutions
- HCD (state) Annual Progress Reports
- Architectural rendering analysis for proposed projects
- Field observation for construction status

**Update frequency:** Weekly during active development; project status fields update continuously as Accela data changes.

**Known limitations:**
- About 12% of projects lack confirmed APN (Assessor Parcel Number) attribution
- Heights for proposed (not-yet-built) projects are taken from latest available plans and may change in design review
- Demolition permits are tracked separately and may not show on the parent project until the permit is linked
- Some 1980s and 1990s completion dates are approximate
- Buildings on UC Berkeley land are tracked but are not subject to city zoning, so their data flows differently than city projects

For a more thorough methodology discussion, see the project repository's `docs/methodology.md`.

## What this notebook will do in the future

### APR-aligned tours

The state Annual Progress Report (APR) is filed each year reporting Berkeley's housing pipeline status to HCD. Future cells will let you generate flyby tours of:

- All projects entitled in a specific calendar year (e.g., entitled in 2020 → flown in 2026 to see what got built vs. stalled)
- All projects with building permits issued in a year
- All projects completed in a year
- All projects with demolition permits issued in a year

Each tour generates a Google Earth `.kml` file on the fly that you click to open. The tour visits each project, hovers, and moves to the next.

This is "walk the APR, watch the buildings" — a visual audit of the table-form report.

- All housing projects within walking distance of a BART station

- Parcels that were rezoned in 2023 versus parcels with new applications since then

This integration is what makes the database useful for journalism: cross-referencing pipeline outcomes with policy and land-use context.

## Notes for journalists, advocates, and researchers

**This data is curated synthesis from public sources.** It's not the City of Berkeley's official record. The official record lives in Accela, in city files, and in HCD filings. This database aggregates and cross-references those sources so you can see patterns the original sources don't make visible on their own.

**For citation:** Cite the underlying sources where possible. This dataset is useful as a starting point and verification tool, not as a primary source.

**To verify a specific project:**
1. Note the address and any permit numbers shown in Google Earth
2. Look up the project in Berkeley's Accela portal at https://aca-prod.accela.com/Berkeley/
3. Cross-check with city documents

**To flag an error:** Open a GitHub issue at https://github.com/blockXblock/berkeley-housing-analysis/issues with the address and what you observed.
